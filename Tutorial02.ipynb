Q4 : Create a random NumPy array of size 6 × 3 and transpose it.

```python
random_array=np.random.rand(6,3)
rand_arr_trans=random_arr.T
print(random_arr)
print(rand_arr_trans)
```

Q5 : Create two random NumPy arrays of size n × p and p × m, and then compute the dot product.

```python

#first random arrray
A_rand=np.random.rand(3,4)
#second random arrray
B_rand=np.random.rand(4,2)
#product array
product_rand=np.dot(A_rand,B_rand)
#print all of them
print(A_rand)
print(B_rand)
print(product_rand)
print(A_rand.shape)
print(B_rand.shape)
print(product_rand.shape)
```
